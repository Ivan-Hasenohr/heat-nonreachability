# Paper figures

Regenerate the five illustrations from the bundled floating snapshots. Outputs and display-scale metadata are written to `../results/figures/`.


```python
from pathlib import Path
import subprocess
import sys
from IPython.display import display, Image

folder = Path.cwd()
if not (folder / "run_paper_figures.py").is_file():
    folder = folder / "figures"
subprocess.run([sys.executable, str(folder / "run_paper_figures.py")], check=True)
output = folder.parent / "results" / "figures"
```
